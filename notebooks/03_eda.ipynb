# 03. Exploratory Data Analysis (EDA)

## Objective
Analyze key sales trends, customer purchasing distributions, top revenue-generating products, and geographic performance using visualizations.

### Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")
pd.set_option('display.float_format', lambda x: '%.2f' % x)
print("Libraries imported.")
```

### Step 2: Load Cleaned Transactions Dataset

```python
df = pd.read_csv('../data/processed/cleaned_transactions.csv')
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
print(f"Loaded {len(df):,} cleaned transaction rows.")
```

### Step 3: Overall Business High-Level KPIs
Calculate total revenue, total order count, unique customer count, and Average Order Value (AOV).

```python
total_revenue = df['Revenue'].sum()
total_orders = df['InvoiceNo'].nunique()
unique_customers = df['CustomerID'].nunique()
aov = total_revenue / total_orders
avg_cust_rev = total_revenue / unique_customers

print(f"Total Revenue:              ${total_revenue:,.2f}")
print(f"Total Unique Orders:        {total_orders:,}")
print(f"Total Unique Customers:     {unique_customers:,}")
print(f"Average Order Value (AOV):  ${aov:,.2f}")
print(f"Avg Revenue / Customer:     ${avg_cust_rev:,.2f}")
```

### Step 4: Monthly Revenue Trend Analysis
Group sales by `InvoiceMonthYear` to observe seasonality and revenue growth over time.

```python
monthly_sales = df.groupby('InvoiceMonthYear').agg(
    Monthly_Revenue=('Revenue', 'sum'),
    Order_Count=('InvoiceNo', 'nunique')
).reset_index()
monthly_sales
```

### Step 5: Visualization — Monthly Revenue Line Chart

```python
plt.figure(figsize=(12, 5))
sns.lineplot(data=monthly_sales, x='InvoiceMonthYear', y='Monthly_Revenue', marker='o', color='#2563EB', linewidth=2.5)
plt.title('Monthly Revenue Performance (2010 - 2011)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Month-Year', fontsize=11)
plt.ylabel('Total Revenue ($)', fontsize=11)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Step 6: Geographic Revenue Analysis (Top Countries)
Calculate total revenue contribution by country.

```python
country_rev = df.groupby('Country').agg(
    Customer_Count=('CustomerID', 'nunique'),
    Order_Count=('InvoiceNo', 'nunique'),
    Total_Revenue=('Revenue', 'sum')
).sort_values(by='Total_Revenue', ascending=False).reset_index()

country_rev['Revenue_Share_%'] = (country_rev['Total_Revenue'] / total_revenue * 100).round(2)
country_rev.head(10)
```

### Step 7: Visualization — Top 10 Countries by Revenue

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=country_rev.head(10), x='Total_Revenue', y='Country', hue='Country', palette='Blues_r', legend=False)
plt.title('Top 10 Countries by Revenue Contribution ($)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Total Revenue ($)', fontsize=11)
plt.ylabel('Country', fontsize=11)
plt.tight_layout()
plt.show()
```

### Step 8: Top 10 Products by Quantity Sold

```python
top_qty_products = df.groupby(['StockCode', 'Description']).agg(
    Total_Quantity=('Quantity', 'sum'),
    Total_Revenue=('Revenue', 'sum')
).sort_values(by='Total_Quantity', ascending=False).head(10).reset_index()
top_qty_products
```

### Step 9: Top 10 Products by Revenue Generated

```python
top_rev_products = df.groupby(['StockCode', 'Description']).agg(
    Total_Revenue=('Revenue', 'sum'),
    Total_Quantity=('Quantity', 'sum')
).sort_values(by='Total_Revenue', ascending=False).head(10).reset_index()
top_rev_products
```

### Step 10: Customer Purchase Frequency Distribution
Analyze how many orders individual customers place.

```python
cust_orders = df.groupby('CustomerID')['InvoiceNo'].nunique()
cust_orders.describe()
```
