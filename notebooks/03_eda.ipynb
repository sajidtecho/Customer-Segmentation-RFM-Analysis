# 03. Exploratory Data Analysis

Analyze revenue trends, top countries, order volume, and top products.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_cleaned = pd.read_csv('../data/processed/cleaned_transactions.csv')
print('Total Revenue: $', round(df_cleaned['Revenue'].sum(), 2))
df_cleaned.groupby('Country')['Revenue'].sum().sort_values(ascending=False).head(10).plot(kind='bar', title='Top Countries by Revenue')
plt.show()
```
