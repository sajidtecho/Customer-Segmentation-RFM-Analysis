# 01. Data Understanding & Initial Exploration

## Objective
The goal of this notebook is to perform initial data understanding on the raw **UCI Online Retail Dataset**. 

We will inspect:
- Dataset dimensions (rows & columns)
- Data types of each attribute
- Missing value counts and percentages
- Duplicate transaction entries
- Summary statistics and initial data quality observations

### Step 1: Import Required Python Libraries
We import `pandas` for data manipulation, `numpy` for numerical operations, and `matplotlib` / `seaborn` for visualization support.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure display options
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', lambda x: '%.2f' % x)
print("Libraries loaded successfully.")
```

### Step 2: Load the Raw Transaction Dataset
We load the raw Excel transaction file located at `../data/raw/Online Retail.xlsx`.

```python
raw_file_path = '../data/raw/Online Retail.xlsx'
df_raw = pd.read_excel(raw_file_path)
print("Raw dataset loaded successfully.")
```

### Step 3: Display First 5 Rows
Inspecting the first few rows helps us understand the structure of individual records.

```python
df_raw.head()
```

### Step 4: Display Last 5 Rows
Checking the bottom rows confirms data consistency through the end of the file.

```python
df_raw.tail()
```

### Step 5: Check Dataset Shape (Rows & Columns)
We determine the total volume of raw transaction records.

```python
rows, cols = df_raw.shape
print(f"Total Rows: {rows:,}")
print(f"Total Columns: {cols}")
```

### Step 6: Inspect Column Names
We verify the exact field names present in the raw file.

```python
df_raw.columns.tolist()
```

### Step 7: Check Data Types & Structure
We inspect data types to identify columns needing type conversions (e.g. InvoiceDate, CustomerID).

```python
df_raw.info()
```

### Step 8: Identify Missing Values per Column
Missing values (especially in `CustomerID` and `Description`) must be identified early.

```python
missing_counts = df_raw.isnull().sum()
missing_counts
```

### Step 9: Calculate Missing Value Percentages
Understanding the proportion of missing data guides our cleaning strategy.

```python
missing_pct = (df_raw.isnull().sum() / len(df_raw) * 100).round(2)
missing_df = pd.DataFrame({'Missing_Count': missing_counts, 'Missing_Percentage': missing_pct})
missing_df
```

### Step 10: Check Duplicate Rows
Identical duplicate rows distort order volumes and total revenue.

```python
duplicate_count = df_raw.duplicated().sum()
print(f"Total Duplicate Rows: {duplicate_count:,}")
```

### Step 11: Inspect Unique Entity Counts
We count unique transactions (`InvoiceNo`), products (`StockCode`), customers (`CustomerID`), and countries.

```python
print(f"Unique Invoices: {df_raw['InvoiceNo'].nunique():,}")
print(f"Unique Products (StockCodes): {df_raw['StockCode'].nunique():,}")
print(f"Unique Customers: {df_raw['CustomerID'].nunique():,}")
print(f"Unique Countries: {df_raw['Country'].nunique():,}")
```

### Step 12: Check Date Range of Transactions
We determine the timeframe covered by the raw dataset.

```python
print(f"Earliest Invoice Date: {df_raw['InvoiceDate'].min()}")
print(f"Latest Invoice Date:   {df_raw['InvoiceDate'].max()}")
```

### Step 13: Summary Descriptive Statistics
We examine summary statistics for numerical attributes (`Quantity`, `UnitPrice`).

```python
df_raw.describe()
```
